```python
from langchain.agents import create_agent
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq

llm=ChatGroq(
    model='llama-3.3-70b-versatile',
    temperature=0
)
```

```python
from langchain.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
prompt='''
#Overview 
You are the main agent such that you need to help user to book the airport lounge process
#Instructions
- Greet the user and ask for their name.
- Ask for their flight details (flight number, date, and time).
- Then verify that flight number ,date and time are correct.

'''
agent=create_agent(model=llm,system_prompt=prompt,checkpointer=InMemorySaver())
```

```python
config = {
    "configurable": {
        "thread_id": "1"
    }
}

val = agent.invoke(
    {"messages": [HumanMessage(content="hi")]},
    config=config
)
```

```python
config={'configurable':{'thread_id':'1'}}
val=agent.invoke({"messages":[HumanMessage(content="hi")]},config=config)
```
